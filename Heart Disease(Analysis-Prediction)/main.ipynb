#### Your dataset contains health-related features. Below is a detailed explanation of each column:

1. **HeartDisease (Binary: 0 or 1)**

0 → The person does not have heart disease.

1 → The person has been diagnosed with heart disease.

**Purpose**: Identifies individuals with or without heart disease.

2. **BMI (Body Mass Index)**

A numeric value representing the BMI, which is calculated as:

𝐵𝑀𝐼 = weight (kg) /height (m)**2

​
 
**Categorization**:

Underweight: BMI < 18.5     =====>     Increased risk of malnutrition, osteoporosis, weak immunity


Normal weight: 18.5 ≤ BMI < 24.9  =====>     Lowest health risks, considered a healthy weight


Overweight: 25 ≤ BMI < 29.9   =====>    Increased risk of cardiovascular diseases, diabetes


Obese: BMI ≥ 30            =====>      Higher risk of severe health conditions (heart disease, stroke, diabetes)


**Purpose**: Determines obesity levels, which impact heart disease risk.


3. **Smoking (Binary: 0 or 1)**

0 → Never smoked.

1 → Smoker (current or past).

**Purpose**: Smoking is a major risk factor for heart disease, stroke, and lung disease.

4. **AlcoholDrinking (Binary: 0 or 1)**

0 → Does not drink alcohol excessively.

1 → Binge drinker (heavy alcohol consumption).

**Purpose**: Excessive alcohol intake increases the risk of heart disease and liver disease.

5. **Stroke (Binary: 0 or 1)**

0 → No history of stroke.

1 → Had a stroke.

**Purpose**: Stroke is a major cardiovascular event linked to high blood pressure and heart disease.

6. **PhysicalHealth (Numeric: 0-30 days)**

Represents the number of days in the past month (30 days) the person experienced poor physical health.

0 → No issues, higher values → More frequent physical health problems.

**Purpose**: Helps measure chronic health conditions affecting well-being.

7. **MentalHealth (Numeric: 0-30 days)**

Represents the number of days in the past month the person experienced poor mental health.

0 → No mental health issues, higher values → More frequent mental health struggles.

**Purpose**: Mental health disorders like depression can impact physical health.

8. **DiffWalking (Binary: 0 or 1)**

0 → No difficulty walking.

1 → Has difficulty walking or climbing stairs.

**Purpose**: Assesses mobility issues, which are linked to disabilities and chronic diseases.

9. **Sex (Categorical: Male or Female)**

Male or Female

**Purpose**: Gender influences health risks, as men and women have different heart disease risk factors.

10. **AgeCategory (Categorical)**

Age groups such as:

18-24

25-29

30-34
...
75-79

80 or older

**Purpose**: Age is a key factor in disease risk, with older individuals having a higher risk of chronic diseases.

11. **Race (Categorical)**

Common categories:

White

Black

Asian

Hispanic

Other

**Purpose**: Some racial/ethnic groups have higher or lower risks for certain diseases due to genetics and lifestyle factors.

12. **Diabetic (Categorical)**

No → Not diabetic.

Yes → Diagnosed with diabetes.

Borderline → Pre-diabetic (at risk).

**Purpose**: Diabetes increases the risk of heart disease, stroke, and kidney failure.

13. **PhysicalActivity (Binary: 0 or 1)**

0 → No regular physical activity.

1 → Engages in physical activity.

**Purpose**: Exercise reduces the risk of obesity, diabetes, and heart disease.

14. **GenHealth (Categorical)**

Self-reported general health status:

Excellent

Very Good

Good

Fair

Poor

**Purpose**: A subjective measure of overall health.

15. **SleepTime (Numeric: 0-24 hours)**

Number of hours of sleep per night.

Recommended sleep: 7-9 hours per night.

**Purpose**: Poor sleep increases stress, obesity, and heart disease risk.

16. **Asthma (Binary: 0 or 1)**

0 → No asthma.

1 → Diagnosed with asthma.

**Purpose**: Asthma can contribute to respiratory issues and affect physical activity levels.

17. **KidneyDisease (Binary: 0 or 1)**

0 → No kidney disease.

1 → Diagnosed with kidney disease.

**Purpose**: Chronic kidney disease is linked to high blood pressure and diabetes.

18. **SkinCancer (Binary: 0 or 1)**

0 → No skin cancer.

1 → Diagnosed with skin cancer.

**Purpose**: Skin cancer can indicate sun exposure habits and overall health risks.

#### 🔹 **Summary**
This dataset covers a wide range of health indicators, from lifestyle factors (e.g., smoking, alcohol use, and physical activity) to chronic conditions (e.g., diabetes, heart disease, kidney disease).** The goal is likely to analyze risk factors for heart disease or other chronic illnesses.

```python

bins = [0, 3, 6, 9, 12, 24]  
labels = ["Very Low (1-3)", "Low (4-6)", "Optimal (7-9)", "High (10-12)", "Excessive (13-24)"]

df["SleepCategory"] = pd.cut(df["SleepTime"], bins=bins, labels=labels, include_lowest=True)


sleep_heart_disease = df.groupby("SleepCategory")["HeartDisease"].value_counts(normalize=True).unstack()


sleep_heart_disease.plot(kind="bar")
plt.title("Proportion of Heart Disease by Sleep Category")
plt.ylabel("Proportion")
plt.xlabel("Sleep Category")
plt.legend(title="Heart Disease", labels=["No", "Yes"])
plt.show()
```

Lack of sleep (less than 6 hours per day):

Increases blood pressure.
Raises cortisol (stress hormone) levels, which leads to inflammation of blood vessels.
Increases insulin resistance, which increases the risk of diabetes and heart disease.
Leads to irregular heartbeats and increases the risk of heart attacks.
Excessive sleep (more than 9 hours per day):

May be linked to heart disease due to lack of physical activity and increased obesity.
Increases the risk of stroke and arterial disease.

#### 🔹According to the American Heart Association (AHA), 7-8 hours of sleep per day is ideal for maintaining heart health.

**General Conclusion**

1️⃣ Strong Link Between Kidney Disease and Heart Disease

A significant proportion of heart disease patients also have kidney disease.

This may be due to kidney disease’s impact on blood pressure and vascular health, increasing cardiovascular risk.

2️⃣ Less Clear but Existing Relationship Between Skin Cancer and Heart Disease

While the connection is not as strong, common factors like age, diet, and immune system effects may play a role.

3️⃣ Asthma and Heart Disease Show a Noticeable Connection


Asthma patients have a higher likelihood of developing heart disease compared to non-asthmatics.

Chronic inflammation and reduced oxygen levels associated with asthma may contribute to cardiovascular issues.

Prevention Strategies

✅ Kidney Disease – Early detection, blood pressure control, and kidney function monitoring can reduce heart disease risk.

✅ Skin Cancer – Maintaining a healthy lifestyle and monitoring cardiovascular health may help manage potential risks.

✅ Asthma – Proper asthma management, avoiding triggers, and regular cardiovascular check-ups are essential preventive measures.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.compose import ColumnTransformer
from sklearn.metrics import accuracy_score,  mean_squared_error,classification_report
```

```python
X = df.drop(columns=['HeartDisease'])  
y = df['HeartDisease']  
```

```python
X
```

```python
y
```

```python
X = pd.get_dummies(X, columns=['AgeCategory', 'Race', 'BMI_Category', 'GenHealth','SleepCategory'], drop_first=True)

```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=20)
```

```python
model = LogisticRegression(max_iter=1500)
```

```python
model.fit(X_train,y_train)
```

```python
model.score(X_train,y_train)
```

```python
model.score(X_test,y_test)
```
